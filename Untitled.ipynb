```python
import pandas as pd

df = pd.read_excel("default of credit card clients.xls", header=1)
df = df.rename(columns={'PAY_0': 'PAY_1', 'default payment next month': 'defpay'})

print(df.shape)
print(df.dtypes.value_counts())
print(df.isna().sum().sum())
print(df['ID'].is_unique)
print(df.drop(columns='ID').duplicated().sum())
```

```python
cols = [c for c in df.columns if c != 'ID']
dups = df[df.duplicated(subset=cols, keep=False)]
print(len(dups))
print(dups.sort_values(cols)[['ID', 'LIMIT_BAL', 'AGE', 'PAY_1', 'BILL_AMT1', 'PAY_AMT1', 'defpay']].head(12))
```

```python
bill = [f'BILL_AMT{i}' for i in range(1, 7)]
pay = [f'PAY_AMT{i}' for i in range(1, 7)]
print((dups[bill + pay].abs().sum(axis=1) == 0).sum())
```

```python
active = dups[dups[bill + pay].abs().sum(axis=1) > 0]
print(active.sort_values(cols)[['ID', 'LIMIT_BAL', 'AGE', 'PAY_1', 'BILL_AMT1', 'PAY_AMT1', 'defpay']])
```
