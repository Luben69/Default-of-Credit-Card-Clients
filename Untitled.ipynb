```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("default of credit card clients.xls", header=1)
df = df.rename(columns={'PAY_0': 'PAY_1', 'default payment next month': 'defpay'})
bill = [f'BILL_AMT{i}' for i in range(1, 7)]
pay = [f'PAY_AMT{i}' for i in range(1, 7)]
num = ['LIMIT_BAL', 'AGE'] + bill + pay
print(df.groupby('defpay')[num].median().T)
```

```python
print(df.shape)
print(df.isnull().sum().sum())
print(df.drop(columns='ID').duplicated().sum())
df.info()
```

```python
counts = df['defpay'].value_counts().sort_index()
pct = df['defpay'].value_counts(normalize=True).sort_index() * 100

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
counts.plot(kind='bar', ax=axes[0], title='Number of customers')
pct.plot(kind='bar', ax=axes[1], title='Percent of customers (%)')
for ax in axes:
    ax.set_xticklabels(['No default (0)', 'Default (1)'], rotation=0)
    ax.bar_label(ax.containers[0])
plt.tight_layout()
plt.show()
```

```python
cols = ['SEX', 'EDUCATION', 'MARRIAGE', 'PAY_1', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
fig, axes = plt.subplots(3, 3, figsize=(18, 11))
for ax, c in zip(axes.ravel(), cols):
    counts = df[c].value_counts().sort_index()
    pct = counts / counts.sum() * 100
    counts.plot(kind='bar', ax=ax, title=c, rot=0)
    ax.bar_label(ax.containers[0], labels=[f'{p:.1f}%' for p in pct], fontsize=8)
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
```

```python
cols = ['SEX', 'EDUCATION', 'MARRIAGE', 'PAY_1', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
fig, axes = plt.subplots(3, 3, figsize=(15, 11))
for ax, c in zip(axes.ravel(), cols):
    df[c].value_counts().sort_index().plot(kind='bar', ax=ax, title=c, rot=0)
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
```

```python
print(df['defpay'].groupby(df['SEX']).value_counts())
print(df['defpay'].groupby(df['SEX']).value_counts(normalize = True))
```

```python
counts = df.groupby('SEX')['defpay'].value_counts().unstack()
pct = df.groupby('SEX')['defpay'].value_counts(normalize=True).unstack() * 100

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
counts.plot(kind='bar', ax=axes[0], title='Customers by sex and default', rot=0)
pct.plot(kind='bar', ax=axes[1], title='Share inside each sex (%)', rot=0)

for cont in axes[0].containers:
    axes[0].bar_label(cont, fmt='%d')
for cont in axes[1].containers:
    axes[1].bar_label(cont, fmt='%.1f')
for ax in axes:
    ax.set_xticklabels(['Male (1)', 'Female (2)'], rotation=0)
    ax.set_xlabel('')
    ax.legend(title='defpay')
plt.tight_layout()
plt.show()
```

```python
print(df['defpay'].groupby(df['EDUCATION']).value_counts())
print(df['defpay'].groupby(df['EDUCATION']).value_counts(normalize = True))
```

```python
def plot_by(col):
    counts = df.groupby(col)['defpay'].value_counts().unstack().fillna(0)
    pct = df.groupby(col)['defpay'].value_counts(normalize=True).unstack().fillna(0) * 100

    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    counts.plot(kind='bar', ax=axes[0], title=f'{col}: customers', rot=0)
    pct.plot(kind='bar', ax=axes[1], title=f'{col}: share inside each code (%)', rot=0)
    for cont in axes[0].containers:
        axes[0].bar_label(cont, fmt='%d', fontsize=8)
    for cont in axes[1].containers:
        axes[1].bar_label(cont, fmt='%.1f', fontsize=8)
    plt.tight_layout()
    plt.show()

plot_by('EDUCATION')
```

```python
print(df['defpay'].groupby(df['MARRIAGE']).value_counts())
print(df['defpay'].groupby(df['MARRIAGE']).value_counts(normalize = True))
```

```python
print(df['defpay'].groupby([df['MARRIAGE'], df['SEX']]).value_counts())
print(df['defpay'].groupby([df['MARRIAGE'], df['SEX']]).value_counts(normalize=True))
```

```python
size = df.groupby(['MARRIAGE', 'SEX']).size().unstack()
rate = df.groupby(['MARRIAGE', 'SEX'])['defpay'].mean().unstack() * 100

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
size.plot(kind='bar', ax=axes[0], title='Customers by marriage and sex', rot=0)
rate.plot(kind='bar', ax=axes[1], title='Default rate (%) by marriage and sex', rot=0)
for cont in axes[0].containers:
    axes[0].bar_label(cont, fmt='%d', fontsize=8)
for cont in axes[1].containers:
    axes[1].bar_label(cont, fmt='%.1f', fontsize=8)
for ax in axes:
    ax.legend(['Male (1)', 'Female (2)'])
    ax.set_xlabel('MARRIAGE')
plt.tight_layout()
plt.show()
```

```python
print(df['defpay'].groupby([df['EDUCATION'], df['SEX']]).value_counts())
print(df['defpay'].groupby([df['EDUCATION'], df['SEX']]).value_counts(normalize=True))
```

```python
def plot_with_sex(col):
    size = df.groupby([col, 'SEX']).size().unstack()
    rate = df.groupby([col, 'SEX'])['defpay'].mean().unstack() * 100

    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    size.plot(kind='bar', ax=axes[0], title=f'Customers by {col} and sex', rot=0)
    rate.plot(kind='bar', ax=axes[1], title=f'Default rate (%) by {col} and sex', rot=0)
    for cont in axes[0].containers:
        axes[0].bar_label(cont, fmt='%d', fontsize=8)
    for cont in axes[1].containers:
        axes[1].bar_label(cont, fmt='%.1f', fontsize=8)
    for ax in axes:
        ax.legend(['Male (1)', 'Female (2)'])
        ax.set_xlabel(col)
    plt.tight_layout()
    plt.show()

plot_with_sex('EDUCATION')
```

```python
df['AgeBin'] = pd.cut(df['AGE'],[20, 25, 30, 35, 40, 50, 60, 80])
print(df['AgeBin'].value_counts().sort_index())
df['defpay'].groupby(df['AgeBin']).value_counts(normalize = True)
```

```python
size = df.groupby('AgeBin', observed=True).size()
rate = df.groupby('AgeBin', observed=True)['defpay'].mean() * 100

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
size.plot(kind='bar', ax=axes[0], title='Customers by age bin', rot=0)
rate.plot(kind='bar', ax=axes[1], title='Default rate (%) by age bin', rot=0)
axes[1].axhline(22.1, color='red', linestyle='--')
axes[0].bar_label(axes[0].containers[0], fmt='%d', fontsize=8)
axes[1].bar_label(axes[1].containers[0], fmt='%.1f', fontsize=8)
for ax in axes:
    ax.set_xlabel('')
plt.tight_layout()
plt.show()
```
