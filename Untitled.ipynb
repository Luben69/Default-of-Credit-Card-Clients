```python
import pandas as pd
df = pd.read_excel("default of credit card clients.xls", header=1)
print(df.shape)
print(df.columns.tolist())
```

```python
df.head(5)
```

```python
print(df['default payment next month'].value_counts())
print(df['default payment next month'].value_counts(normalize=True))
```

```python
print(df['EDUCATION'].value_counts().sort_index())
print(df['MARRIAGE'].value_counts().sort_index())
print(df['PAY_0'].value_counts().sort_index())
```

```python
print(df.groupby('PAY_0')[['BILL_AMT1', 'PAY_AMT1']].median())
```

```python
print(df['BILL_AMT1'].agg(['mean', 'median', 'max']))
```

```python
print((df['BILL_AMT1'] <= 0).groupby(df['PAY_0']).mean())
```

```python
sub = df[df['PAY_0'].isin([-2, 1])]
print(sub.groupby('PAY_0')['BILL_AMT1'].agg(
    zero=lambda s: (s == 0).sum(),
    negative=lambda s: (s < 0).sum(),
    total='size'
))
```

```python
for col in ['EDUCATION', 'MARRIAGE', 'PAY_0']:
    print(f"\n{col}")
    print("Missing:", df[col].isna().sum())
    print("Unique values:", sorted(df[col].unique()))
```
